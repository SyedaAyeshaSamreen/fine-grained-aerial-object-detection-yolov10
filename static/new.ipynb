```python
from ultralytics import YOLO
import cv2
import os
from pathlib import Path
```

```python
# Set dataset directory path
dataset_dir = "D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY - Copy/CODE/train/images"

# Example structure
train_images = os.path.join(dataset_dir, "train/images")
val_images = os.path.join(dataset_dir, "valid/images")
train_labels = os.path.join(dataset_dir, "train/labels")
val_labels = os.path.join(dataset_dir, "valid/labels")
```

```python
# Create the YAML file for the dataset
yaml_content = """
train: D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY - Copy/CODE/train/images
val: D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY - Copy/CODE/valid/images
test: D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY - Copy/CODE/test/images

nc: 18
names: ['airport', 'baseball-diamond', 'basketball-court', 'bridge', 'container-crane', 'ground-track-field', 'harbor', 'helicopter', 'helipad', 'large-vehicle', 'plane', 'roundabout', 'ship', 'small-vehicle', 'soccer-ball-field', 'storage-tank', 'swimming-pool', 'tennis-court']

"""

# Save the YAML file
with open("data.yaml", "w") as f:
    f.write(yaml_content)
```

```python
# Initialize YOLOv8 model (e.g., yolov8n for Nano version)
model = YOLO('yolov10m.pt')  # Or load a pretrained model like 'yolov8n.pt'
```

```python
# Train the YOLOv8 model
model.train(data='data.yaml', epochs=7, imgsz=640, batch=16)
```

```python
# Evaluate the trained model on validation data
metrics = model.val()
print(metrics)
```

```python
#Image code


from ultralytics import YOLO
import cv2

# Load YOLO model
model = YOLO(r'D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY/CODE/runs/detect/train2/weights/best.pt')

# Input and output image paths
input_image_path = r"D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY/CODE/test/images/000096_jpg.rf.cedc5e4d3b2cbd3d9163b242f21fdd3b.jpg"
output_image_path = r"D:/SHIVAA/CODES/OBJECTS_IN_AIRPORT_RUNWAY/CODE/output.jpg"

# Load the image
image = cv2.imread(input_image_path)

# Perform detection
results = model.predict(image)

# Annotate detections
for result in results[0].boxes.data.tolist():
    x1, y1, x2, y2, conf, cls = result
    x1, y1, x2, y2 = map(int, (x1, y1, x2, y2))  # Bounding box coordinates
    label = f"{model.names[int(cls)]} {conf:.2f}"  # Label with confidence

    # Draw bounding box and label
    cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)  # Green box
    cv2.putText(image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

# Save the output image
cv2.imwrite(output_image_path, image)


```
